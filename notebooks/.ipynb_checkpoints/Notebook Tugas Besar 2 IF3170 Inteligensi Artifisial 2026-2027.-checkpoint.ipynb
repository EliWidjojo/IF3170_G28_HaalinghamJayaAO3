### Data Errors

This part covers wrong types, inconsistent labels, and impossible values.

```python
non_negative_cols = ["num_adults", "num_children", "weekend_nights", "weekday_nights", "days_before_arrival", "prior_cancellations", "prior_completed_bookings", "nightly_rate_idr", "special_requests"]

def correct_data_errors(frame):
    frame = frame.copy()

    # Membersihkan whitespace kategori
    categorical = frame.select_dtypes(include=["object", "category"]).columns
    for col in categorical:
        frame[col] = frame[col].map(lambda v: v.strip() if isinstance(v, str) else v)

    # Konversi kolom numerik
    numeric_candidates = ["num_adults", "num_children", "weekend_nights", "weekday_nights", "needs_parking", "days_before_arrival", "prior_cancellations", "prior_completed_bookings", "nightly_rate_idr", "special_requests", "arrival_year", "arrival_month", "arrival_day"]
    for col in numeric_candidates:
        if col in frame.columns:
            frame[col] = pd.to_numeric(frame[col], errors="coerce")

    # Nilai negatif -> NaN
    for col in non_negative_cols:
        if col in frame.columns:
            frame.loc[frame[col] < 0, col] = np.nan

    # Validasi binary
    for col in ["needs_parking", "is_returning_guest"]:
        if col in frame.columns:
            frame.loc[~frame[col].isin([0, 1]) & frame[col].notna(), col] = np.nan

    # Validasi rentang tanggal
    if "arrival_month" in frame.columns:
        frame.loc[~frame["arrival_month"].between(1, 12), "arrival_month"] = np.nan
    if "arrival_day" in frame.columns:
        frame.loc[~frame["arrival_day"].between(1, 31), "arrival_day"] = np.nan

    # num_adults = 0 tetap dipertahankan.
    return frame

# Terapkan
X_train = correct_data_errors(X_train)
X_val = correct_data_errors(X_val)
X_test = correct_data_errors(X_test)

print("Missing values setelah cleaning:")
display(X_train.isna().sum().loc[lambda x: x > 0].to_frame("missing_train"))
```

> 1. Dilakukan pembersihan whitespace (spasi berlebih) pada seluruh fitur kategorikal untuk mencegah kategori ganda (misal: "Breakfast" vs "Breakfast ").

> 2. Daftar kolom yang diperiksa untuk nilai negatif disesuaikan dengan skema dataset kompetisi. Nilai negatif (jika ada) akan diubah menjadi NaN agar ditangani oleh Imputer.

> 3. Penanganan Nilai Mustahil (Impossible Values): Ditemukan 106 baris dengan num_adults == 0. Secara logika bisnis, reservasi hotel tidak mungkin dilakukan tanpa orang dewasa. Oleh karena itu, nilai 0 pada num_adults diubah menjadi NaN agar dapat diimputasi dengan nilai median dari data yang valid, sehingga model tidak belajar dari pola yang salah.

### Outliers

This part covers values far from the rest of the data.

```python
continuous_for_iqr = ["nightly_rate_idr", "days_before_arrival"]
outlier_summary = []

for col in continuous_for_iqr:
    if col not in X_train.columns: continue
    q1 = X_train[col].quantile(0.25)
    q3 = X_train[col].quantile(0.75)
    iqr = q3 - q1
    lower = q1 - CFG.OUTLIER_IQR_FACTOR * iqr
    upper = q3 + CFG.OUTLIER_IQR_FACTOR * iqr
    n_outliers = ((X_train[col] < lower) | (X_train[col] > upper)).sum()

    outlier_summary.append({
        "feature": col, "Q1": q1, "Q3": q3, "IQR": iqr,
        "lower_bound": lower, "upper_bound": upper,
        "n_outliers": n_outliers, "outlier_pct": (n_outliers / len(X_train) * 100)
    })

display(pd.DataFrame(outlier_summary))
# Nilai outlier DIPERTAHANKAN (tidak di-clip)
```

### Imputasi

```python
for df in [X_train, X_val, X_test]:
    df["special_requests_missing"] = (
        df["special_requests"].isna().astype(int)
    )

print("Missing indicator berhasil dibuat.")
display(
    X_train[["special_requests", "special_requests_missing"]].head(3)
)
```

```python
# Ambil daftar kolom dinamis (berdasarkan NUM_COLS dan CAT_COLS dari atas)
num_cols_to_impute = [c for c in NUM_COLS if c in X_train.columns]
cat_cols_to_impute = [c for c in CAT_COLS if c in X_train.columns]

# Imputasi Numerik (Median)
numeric_imputer = SimpleImputer(strategy="median")
X_train[num_cols_to_impute] = numeric_imputer.fit_transform(X_train[num_cols_to_impute])
X_val[num_cols_to_impute] = numeric_imputer.transform(X_val[num_cols_to_impute])
X_test[num_cols_to_impute] = numeric_imputer.transform(X_test[num_cols_to_impute])

# Imputasi Kategorikal (Modus)
categorical_imputer = SimpleImputer(strategy="most_frequent")
X_train[cat_cols_to_impute] = categorical_imputer.fit_transform(X_train[cat_cols_to_impute])
X_val[cat_cols_to_impute] = categorical_imputer.transform(X_val[cat_cols_to_impute])
X_test[cat_cols_to_impute] = categorical_imputer.transform(X_test[cat_cols_to_impute])

# Pastikan tidak ada missing lagi (kecuali special_requests_missing yg memang 0/1)
assert not X_train[num_cols_to_impute + cat_cols_to_impute].isna().any().any(), "Masih ada missing!"
print("Imputasi selesai.")
```

### Attribute Construction

<!-- rubric v3: derived columns, rate_per_guest_night -->

This part covers new attributes derived from existing ones.

```python
def engineer_features(df):
    df = df.copy()

    # 1. Fitur agregat & interaksi
    df["total_guests"] = df["num_adults"] + df["num_children"]
    df["total_nights"] = df["weekend_nights"] + df["weekday_nights"]
    df["total_cost"] = df["nightly_rate_idr"] * df["total_nights"]
    df["lead_time_per_request"] = df["days_before_arrival"] / (df["num_special_requests"] + 1.0)
    
    # 2. Fitur historis (pakai rumus smoothing Anda, jauh lebih stabil dari rasio mentah)
    df["historical_cancel_rate"] = (
        (df["prior_cancellations"] + 1) / 
        (df["prior_cancellations"] + df["prior_completed_bookings"] + 2)
    )

    # 3. Transformasi Log untuk meredam outlier ekstrem (PENTING untuk KNN)
    df["log_days_before_arrival"] = np.log1p(df["days_before_arrival"].clip(lower=0))
    df["log_nightly_rate"] = np.log1p(df["nightly_rate_idr"].clip(lower=0))
    df["log_total_cost"] = np.log1p(df["total_cost"].clip(lower=0))

    # 4. Representasi siklik bulan
    month = pd.to_numeric(df["arrival_month"], errors="coerce")
    df["arrival_month_sin"] = np.sin(2 * np.pi * month / 12)
    df["arrival_month_cos"] = np.cos(2 * np.pi * month / 12)

    # 5. Hari dalam minggu berdasarkan tanggal kedatangan
    date = pd.to_datetime(
        df[["arrival_year", "arrival_month", "arrival_day"]].rename(
            columns={
                "arrival_year": "year",
                "arrival_month": "month",
                "arrival_day": "day"
            }
        ),
        errors="coerce"
    )
    df["day_of_week"] = date.dt.dayofweek

    return df

# Terapkan Feature Engineering
X_train = engineer_features(X_train)
X_val = engineer_features(X_val)
X_test = engineer_features(X_test)

# Imputasi day_of_week menggunakan modus training
mode_dow = X_train["day_of_week"].mode().iloc[0]

for df in [X_train, X_val, X_test]:
    dow = df["day_of_week"].fillna(mode_dow)
    
    # Ubah day_of_week menjadi siklikal agar Minggu (6) & Senin (0) berdekatan
    df["dow_sin"] = np.sin(2 * np.pi * dow / 7)
    df["dow_cos"] = np.cos(2 * np.pi * dow / 7)

    # Hapus kolom tanggal mentah & day_of_week linear agar tidak menambah noise jarak
    cols_to_drop = ["arrival_year", "arrival_month", "arrival_day", "day_of_week", "booking_id"]
    df.drop(columns=[c for c in cols_to_drop if c in df.columns], inplace=True)
```

### Log Transformation

```python
# Simpan salinan SEBELUM log transform untuk Categorical NB
X_train_nb_raw = X_train.copy()
X_val_nb_raw = X_val.copy()
X_test_nb_raw = X_test.copy()

# Terapkan Log-Transform (hanya pada fitur yang sangat skewed)
LOG_COLS = ["prior_cancellations", "prior_completed_bookings"]

def apply_log_transform(frame, columns):
    frame = frame.copy()
    for col in columns:
        if col in frame.columns:
            frame[col] = np.log1p(frame[col])
    return frame

X_train = apply_log_transform(X_train, LOG_COLS)
X_val = apply_log_transform(X_val, LOG_COLS)
X_test = apply_log_transform(X_test, LOG_COLS)

print("Log-transform selesai diterapkan pada fitur:", LOG_COLS)
```

### Feature Scaling

This part covers normalization to the range 0 to 1 and standardization.

```python
# Untuk SVM, LR
numeric_transformer_scaled = Pipeline(steps=[("scaler", RobustScaler())])

# Untuk Decision Tree (tanpa scaling)
numeric_transformer_unscaled = "passthrough"

# Untuk KNN
ordinal_cols = ["meal_package"] 
onehot_cols = ["room_category", "booking_channel"] 
num_cols = [c for c in X_train.columns if c not in ordinal_cols + onehot_cols]

numeric_transformer_knn = Pipeline(steps=[
    ("imputer", SimpleImputer(strategy="median")),
    ("robust", RobustScaler()),
    ("minmax", MinMaxScaler()) # Menyamakan rentang agar seimbang dengan OneHot (0-1)
])

ordinal_transformer_knn = Pipeline(steps=[
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("ordinal", OrdinalEncoder(handle_unknown="use_encoded_value", unknown_value=-1)),
    ("minmax", MinMaxScaler())
])

onehot_transformer_knn = Pipeline(steps=[
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("onehot", OneHotEncoder(drop="first", sparse_output=False, handle_unknown="ignore"))
])

preprocessor_knn = ColumnTransformer(
    transformers=[
        ("num", numeric_transformer_knn, num_cols),
        ("ord", ordinal_transformer_knn, ordinal_cols),
        ("cat", onehot_transformer_knn, onehot_cols)
    ]
)

X_train_knn = preprocessor_knn.fit_transform(X_train)
X_val_knn = preprocessor_knn.transform(X_val)
X_test_knn = preprocessor_knn.transform(X_test)
```

> Imputasi data numerik menggunakan median untuk semua fitur numerik karena tahan terhadap outlier dan lebih masuk akal untuk fitur count (misal: imputasi 2 orang dewasa lebih wajar daripada 1.84 orang). Namun, khusus Naive Bayes fitur numerik diubah menjadi data kategorikal (diskrit) menggunakan KBinsDiscretizer karena Naive Bayes Categorical tidak dapat memproses angka desimal kontinu (hasil scaling). Scaling menggunakan RobustScaler karena data kita banyak mengandung outlier ekstrem dan right-skewed (seperti nightly_rate_idr dan days_before_arrival), RobustScaler yang berbasis Median dan IQR akan menormalkan skala tanpa tertarik oleh nilai ekstrem tersebut.

```python
important_cols = ["log_days_before_arrival", "num_special_requests", "historical_cancel_rate", "log_nightly_rate"]
for col in important_cols:
    if col in num_cols:
        idx = num_cols.index(col)
        X_train_knn[:, idx] *= 1.8
        X_val_knn[:, idx] *= 1.8
        X_test_knn[:, idx] *= 1.8
```

# Preparation Pipeline <a name="6"></a>

---

<!-- rubric v3: fitted steps, fitted_prep_steps -->

The pipeline combines the preparation steps so they run in one consistent order.

```python
# Kelompok fitur biner
BINARY_FEATURES = [
    "needs_parking",
    "is_returning_guest",
    "special_requests_missing"
]

# Fitur numerik
NUM_CONTINUOUS = [
    c for c in NUM_COLS
    if c in X_train.columns
] + [
    "total_guests",
    "total_nights",
    "total_cost",
    "historical_cancel_rate",
    "arrival_month_sin",
    "arrival_month_cos"
]

# Fitur kategorikal
CATEGORICAL_FEATURES = [
    c for c in CAT_COLS
    if c in X_train.columns
    and c not in BINARY_FEATURES
    and c not in ["arrival_month", "arrival_day"]
]

if "day_of_week" in X_train.columns:
    CATEGORICAL_FEATURES.append("day_of_week")

# Validasi pembagian kelompok fitur
assert len(NUM_CONTINUOUS) == len(set(NUM_CONTINUOUS))
assert len(BINARY_FEATURES) == len(set(BINARY_FEATURES))
assert len(CATEGORICAL_FEATURES) == len(set(CATEGORICAL_FEATURES))

assert set(NUM_CONTINUOUS).isdisjoint(BINARY_FEATURES)
assert set(NUM_CONTINUOUS).isdisjoint(CATEGORICAL_FEATURES)
assert set(BINARY_FEATURES).isdisjoint(CATEGORICAL_FEATURES)

ALL_FEATURES = (
    set(NUM_CONTINUOUS)
    | set(BINARY_FEATURES)
    | set(CATEGORICAL_FEATURES)
)

assert ALL_FEATURES == set(X_train.columns), (
    f"Fitur tidak sinkron. "
    f"Belum terkelompok: {set(X_train.columns) - ALL_FEATURES}. "
    f"Fitur berlebih: {ALL_FEATURES - set(X_train.columns)}"
)

# Preprocessing untuk KNN, SVM, Logistic Regression
preprocessor_scaled = ColumnTransformer(
    transformers=[
        ("num", numeric_transformer_scaled, NUM_CONTINUOUS),
        ("binary", "passthrough", BINARY_FEATURES),
        ("cat", categorical_transformer, CATEGORICAL_FEATURES)
    ],
    remainder="drop",
    verbose_feature_names_out=False
)

# Preprocessing untuk Decision Tree CART
preprocessor_unscaled = ColumnTransformer(
    transformers=[
        ("num", "passthrough", NUM_CONTINUOUS),
        ("binary", "passthrough", BINARY_FEATURES),
        ("cat", categorical_transformer, CATEGORICAL_FEATURES)
    ],
    remainder="drop",
    verbose_feature_names_out=False
)

# Fit hanya pada training
X_train_scaled = preprocessor_scaled.fit_transform(X_train)
X_val_scaled = preprocessor_scaled.transform(X_val)
X_test_scaled = preprocessor_scaled.transform(X_test)

X_train_unscaled = preprocessor_unscaled.fit_transform(X_train)
X_val_unscaled = preprocessor_unscaled.transform(X_val)
X_test_unscaled = preprocessor_unscaled.transform(X_test)

# Pemeriksaan dimensi dan nilai
for name, matrices in {
    "Scaled": (X_train_scaled, X_val_scaled, X_test_scaled),
    "Unscaled": (X_train_unscaled, X_val_unscaled, X_test_unscaled)
}.items():

    train_matrix, val_matrix, test_matrix = matrices

    assert train_matrix.shape[1] == val_matrix.shape[1]
    assert train_matrix.shape[1] == test_matrix.shape[1]

    for matrix in matrices:
      matrix_numeric = np.asarray(matrix, dtype=np.float64)

      assert np.isfinite(matrix_numeric).all(), (
          f"{name} mengandung NaN atau infinity"
      )

    print(
        f"{name}: Train {train_matrix.shape}, "
        f"Val {val_matrix.shape}, "
        f"Test {test_matrix.shape}"
    )

print("Preparation Pipeline selesai.")
```

```python

# Categorical Naive Bayes Preprocessing

# Fitur hitungan yang menggunakan kategori khusus
NB_COUNT_FEATURES = [
    "num_children",
    "prior_cancellations",
    "prior_completed_bookings"
]

# Fitur numerik yang menggunakan quantile discretization
NB_NUMERIC_FEATURES = [
    col for col in NUM_CONTINUOUS
    if col not in NB_COUNT_FEATURES
]

# Fitur kategorikal dan biner
NB_CATEGORICAL_FEATURES = (
    CATEGORICAL_FEATURES + BINARY_FEATURES
)

# Validasi bahwa semua fitur terkelompok
NB_ALL_FEATURES = (
    NB_COUNT_FEATURES
    + NB_NUMERIC_FEATURES
    + NB_CATEGORICAL_FEATURES
)

assert len(NB_ALL_FEATURES) == len(set(NB_ALL_FEATURES))
assert set(NB_ALL_FEATURES) == set(X_train_nb_raw.columns)


# 1. Diskretisasi fitur numerik berdasarkan kuantil training
def fit_nb_discretization(df, columns, n_bins=5):
    bin_edges = {}

    for col in columns:
        quantiles = df[col].quantile(
            np.linspace(0, 1, n_bins + 1)
        ).to_numpy(dtype=float)

        edges = np.unique(quantiles)

        if len(edges) < 2:
            edges = np.array([-np.inf, np.inf])
        else:
            edges[0] = -np.inf
            edges[-1] = np.inf

        bin_edges[col] = edges

    return bin_edges


def apply_nb_discretization(df, columns, bin_edges):
    result = pd.DataFrame(index=df.index)

    for col in columns:
        bins = pd.cut(
            df[col],
            bins=bin_edges[col],
            labels=False,
            include_lowest=True
        )

        if bins.isna().any():
            raise ValueError(
                f"Kolom {col} menghasilkan missing saat binning"
            )

        result[col] = bins.map(
            lambda x: f"bin_{int(x)}"
        )

    return result


# 2. Diskretisasi khusus fitur hitungan
def count_to_category(series):
    if series.isna().any():
        raise ValueError("Fitur hitungan masih memiliki NaN")

    if (series < 0).any():
        raise ValueError("Fitur hitungan memiliki nilai negatif")

    return series.apply(
        lambda x:
            "count_0" if x == 0
            else "count_1" if x == 1
            else "count_2" if x == 2
            else "count_3plus"
    )


def apply_count_discretization(df, columns):
    result = pd.DataFrame(index=df.index)

    for col in columns:
        result[col] = count_to_category(df[col])

    return result


# 3. Fit batas bin hanya dari training
nb_bin_edges = fit_nb_discretization(
    X_train_nb_raw,
    NB_NUMERIC_FEATURES,
    n_bins=5
)


# 4. Transform fitur numerik
X_train_nb_num = apply_nb_discretization(
    X_train_nb_raw, NB_NUMERIC_FEATURES, nb_bin_edges
)

X_val_nb_num = apply_nb_discretization(
    X_val_nb_raw, NB_NUMERIC_FEATURES, nb_bin_edges
)

X_test_nb_num = apply_nb_discretization(
    X_test_nb_raw, NB_NUMERIC_FEATURES, nb_bin_edges
)


# 5. Transform fitur hitungan
X_train_nb_count = apply_count_discretization(
    X_train_nb_raw, NB_COUNT_FEATURES
)

X_val_nb_count = apply_count_discretization(
    X_val_nb_raw, NB_COUNT_FEATURES
)

X_test_nb_count = apply_count_discretization(
    X_test_nb_raw, NB_COUNT_FEATURES
)


# 6. Ambil fitur kategorikal dan biner
X_train_nb_cat = X_train_nb_raw[
    NB_CATEGORICAL_FEATURES
].copy()

X_val_nb_cat = X_val_nb_raw[
    NB_CATEGORICAL_FEATURES
].copy()

X_test_nb_cat = X_test_nb_raw[
    NB_CATEGORICAL_FEATURES
].copy()


# 7. Gabungkan seluruh fitur
def combine_nb_features(numeric, count, categorical):
    result = pd.concat(
        [numeric, count, categorical],
        axis=1
    )
    return result.astype(str)


X_train_nb = combine_nb_features(
    X_train_nb_num,
    X_train_nb_count,
    X_train_nb_cat
)

X_val_nb = combine_nb_features(
    X_val_nb_num,
    X_val_nb_count,
    X_val_nb_cat
)

X_test_nb = combine_nb_features(
    X_test_nb_num,
    X_test_nb_count,
    X_test_nb_cat
)


# 8. Samakan urutan fitur
X_val_nb = X_val_nb.reindex(columns=X_train_nb.columns)
X_test_nb = X_test_nb.reindex(columns=X_train_nb.columns)


# 9. Validasi hasil preprocessing
for name, df in [
    ("Training", X_train_nb),
    ("Validation", X_val_nb),
    ("Test", X_test_nb)
]:
    assert df.shape[1] == X_train_nb.shape[1]
    assert df.columns.tolist() == X_train_nb.columns.tolist()
    assert not df.isna().any().any()

    print(f"{name}: {df.shape}")


# 10. Periksa jumlah kategori tiap fitur numerik
print("\nJumlah kategori fitur numerik:")

for col in NB_NUMERIC_FEATURES + NB_COUNT_FEATURES:
    print(
        f"{col}: {X_train_nb[col].nunique()} kategori"
    )


# 11. Distribusi fitur hitungan
print("\nDistribusi fitur hitungan:")

for col in NB_COUNT_FEATURES:
    print(f"\n{col}")
    print(X_train_nb[col].value_counts().sort_index())


# 12. Periksa unseen category pada validation dan test
print("\nKategori unseen pada validation dan test:")

for name, df in [
    ("Validation", X_val_nb),
    ("Test", X_test_nb)
]:
    unseen_counts = {
        col: int(
            (~df[col].isin(X_train_nb[col].unique())).sum()
        )
        for col in X_train_nb.columns
    }

    print(
        name,
        {
            k: v for k, v in unseen_counts.items()
            if v > 0
        }
    )


print("\nCategorical NB preprocessing selesai!")
display(X_train_nb.head())
```

```python
# Pisahkan baseline dari model hasil tuning
baseline_models = {}
baseline_predictions = {}
baseline_results = []

# Dictionary untuk tahap tuning berikutnya
best_models = {}
best_predictions = {}


def evaluate_model(y_true, y_pred):
    return {
        "Accuracy": accuracy_score(y_true, y_pred),
        "Precision": precision_score(
            y_true, y_pred,
            average="macro", zero_division=0
        ),
        "Recall": recall_score(
            y_true, y_pred,
            average="macro", zero_division=0
        ),
        "F1-Score": f1_score(
            y_true, y_pred,
            average="macro", zero_division=0
        )
    }


def evaluate_baseline(name, model, X_train_data, X_val_data):
    y_tr = np.asarray(y_train).ravel()
    y_va = np.asarray(y_val).ravel()

    assert len(X_train_data) == len(y_tr)
    assert len(X_val_data) == len(y_va)
    assert X_train_data.shape[1] == X_val_data.shape[1]

    print(f"\n=== Baseline {name} ===")

    start_time = time.perf_counter()

    model.fit(X_train_data, y_tr)

    fit_time = time.perf_counter() - start_time

    start_predict = time.perf_counter()

    y_pred = model.predict(X_val_data)

    predict_time = time.perf_counter() - start_predict

    assert len(y_pred) == len(y_va)
    assert np.isin(y_pred, np.unique(y_tr)).all()

    metrics = evaluate_model(y_va, y_pred)

    baseline_models[name] = model
    baseline_predictions[name] = y_pred

    baseline_results.append({
        "Model": name,
        **metrics,
        "Fit Time (s)": fit_time,
        "Predict Time (s)": predict_time
    })

    print("\nEvaluation Metrics:")
    display(pd.Series(metrics, name=name).to_frame())

    print("\nClassification Report:")
    print(
        classification_report(
            y_va, y_pred,
            labels=[0, 1],
            zero_division=0
        )
    )

    print("\nConfusion Matrix:")
    display(
        pd.DataFrame(
            confusion_matrix(y_va, y_pred, labels=[0, 1]),
            index=["Actual 0", "Actual 1"],
            columns=["Predicted 0", "Predicted 1"]
        )
    )

    print(f"\nFit time: {fit_time:.2f} detik")
    print(f"Predict time: {predict_time:.2f} detik")

    return model, y_pred, metrics
```

## K-Nearest Neighbors

## Improvements (Optional)

This part covers hyperparameter tuning, cross-validation, and further visualization.

```python
"""Model improvement untuk lima estimator IF3170 from scratch.

Rujukan utama (sesuai daftar pustaka tugas): [4], [5], [10], [15]-[21], [23], [26].
CATATAN: data input dalam notebook ini telah dipreproses sebelum inner split.
Untuk evaluasi final tanpa preprocessing leakage, fit ulang seluruh tahap preprocessing
pada inner-training tiap split/fold. Jangan gunakan holdout final untuk tuning ulang.

Prasyarat dalam notebook: KNN, DecisionTreeCART, LogisticReg, SVM, NaiveBayes,
X_train_scaled, X_val_scaled, X_train_unscaled, X_val_unscaled,
X_train_nb, X_val_nb, y_train, y_val, CFG.SEED.
"""
import time
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, classification_report, confusion_matrix

# JANGAN ubah validation final untuk memilih parameter.
seed = CFG.SEED
Y = np.asarray(y_train).ravel()
Y_VAL = np.asarray(y_val).ravel()
inner_train_idx, inner_val_idx = train_test_split(
    np.arange(len(Y)), test_size=0.20, stratify=Y, random_state=seed
)

MODEL_INPUTS = {
    'KNN': (np.asarray(X_train_scaled, dtype=np.float64), np.asarray(X_val_scaled, dtype=np.float64)),
    'CART': (np.asarray(X_train_unscaled, dtype=np.float64), np.asarray(X_val_unscaled, dtype=np.float64)),
    'Logistic Regression': (np.asarray(X_train_scaled, dtype=np.float64), np.asarray(X_val_scaled, dtype=np.float64)),
    'SVM': (np.asarray(X_train_scaled, dtype=np.float64), np.asarray(X_val_scaled, dtype=np.float64)),
    'Naive Bayes': (X_train_nb.reset_index(drop=True), X_val_nb.reset_index(drop=True)),
}

# Grid kecil agar bisa dijalankan dahulu. Silakan perluas SETELAH mengecek biaya komputasi.
SEARCH_SPACE = {
    'KNN': [
        {'k': 5, 'distance_metric': 'euclidean', 'weights': 'uniform'},
        {'k': 9, 'distance_metric': 'euclidean', 'weights': 'distance'},
        {'k': 11, 'distance_metric': 'manhattan', 'weights': 'distance'},
    ],
    'CART': [
        {'max_depth': 4, 'min_samples_split': 10, 'min_samples_leaf': 4},
        {'max_depth': 6, 'min_samples_split': 10, 'min_samples_leaf': 4},
        {'max_depth': 8, 'min_samples_split': 20, 'min_samples_leaf': 8},
    ],
    'Logistic Regression': [
        {'learning_rate': 0.01, 'n_epochs': 200, 'penalty': None, 'lambda1': 0., 'lambda2': 0.},
        {'learning_rate': 0.01, 'n_epochs': 200, 'penalty': 'l1', 'lambda1': 0.1, 'lambda2': 0.},
        {'learning_rate': 0.01, 'n_epochs': 200, 'penalty': 'l2', 'lambda1': 0., 'lambda2': 0.1},
        {'learning_rate': 0.01, 'n_epochs': 200, 'penalty': 'elastic', 'lambda1': 0.1, 'lambda2': 0.1},
    ],
    'SVM': [
        {'kernel': 'linear', 'C': 0.5, 'lr': 0.01, 'epochs': 50},
        {'kernel': 'linear', 'C': 2.0, 'lr': 0.01, 'epochs': 50},
        {'kernel': 'rbf', 'C': 1.0, 'lr': 0.005, 'epochs': 50, 'max_kernel_samples': 2000},
        {'kernel': 'poly', 'C': 1.0, 'degree': 2, 'lr': 0.005, 'epochs': 50, 'max_kernel_samples': 2000},
    ],
    'Naive Bayes': [{'alpha': a} for a in (0.5, 1.0, 2.0)],
}


def build_model(name, cfg):
    if name == 'KNN':
        return KNN(**cfg)
    if name == 'CART':
        return DecisionTreeCART(**cfg)
    if name == 'Logistic Regression':
        return LogisticReg(**cfg, batch_size=256, shuffle=True, random_state=seed)
    if name == 'SVM':
        return SVM(**cfg, gamma='scale', random_state=seed)
    if name == 'Naive Bayes':
        return NaiveBayes(**cfg)
    raise ValueError(name)


def subset(X, idx):
    return X.iloc[idx] if isinstance(X, pd.DataFrame) else X[idx]


def scores(y, pred):
    return {
        'Accuracy': accuracy_score(y, pred),
        'Macro Precision': precision_score(y, pred, average='macro', zero_division=0),
        'Macro Recall': recall_score(y, pred, average='macro', zero_division=0),
        'Macro F1': f1_score(y, pred, average='macro', zero_division=0),
        'Recall Canceled': recall_score(y, pred, pos_label=1, zero_division=0),
    }

# 1. Pilih parameter dari inner validation SAJA
improvement_tuning_rows = []
improvement_best_params = {}
for name, configurations in SEARCH_SPACE.items():
    X_train_all, _ = MODEL_INPUTS[name]
    best_f1 = -1.0
    best_cfg = None
    print(f'\n=== Tuning {name}: {len(configurations)} konfigurasi ===')
    for cfg in configurations:
        model = build_model(name, cfg)
        started = time.perf_counter()
        model.fit(subset(X_train_all, inner_train_idx), Y[inner_train_idx])
        pred = model.predict(subset(X_train_all, inner_val_idx))
        duration = time.perf_counter() - started
        metric = scores(Y[inner_val_idx], pred)
        improvement_tuning_rows.append({
            'Model': name, 'Parameters': str(cfg), **metric, 'Time (s)': round(duration, 2)
        })
        print(f'  {cfg} -> inner Macro F1={metric["Macro F1"]:.4f} ({duration:.1f}s)')
        if metric['Macro F1'] > best_f1:
            best_f1, best_cfg = metric['Macro F1'], cfg.copy()
    improvement_best_params[name] = best_cfg

improvement_tuning_df = pd.DataFrame(improvement_tuning_rows)
print('\n=== Hasil inner tuning ===')
display(improvement_tuning_df.sort_values(['Model', 'Macro F1'], ascending=[True, False]))

# 2. Refit model terpilih memakai SEMUA training, kemudian nilai pada final validation sekali.
improvement_models = {}
improvement_predictions = {}
improvement_final_rows = []
for name, cfg in improvement_best_params.items():
    X_tr, X_va = MODEL_INPUTS[name]
    model = build_model(name, cfg)
    started = time.perf_counter()
    model.fit(X_tr, Y)
    fit_seconds = time.perf_counter() - started
    pred = model.predict(X_va)
    metric = scores(Y_VAL, pred)
    improvement_models[name] = model
    improvement_predictions[name] = pred
    improvement_final_rows.append({
        'Model': name, 'Best Parameters': str(cfg), **metric, 'Fit Time (s)': round(fit_seconds, 2)
    })
    print(f'\n{name} | final validation Macro F1={metric["Macro F1"]:.4f}')
    print(classification_report(Y_VAL, pred, labels=[0, 1], zero_division=0))
    print(pd.DataFrame(confusion_matrix(Y_VAL, pred, labels=[0, 1]),
                       index=['Actual 0', 'Actual 1'], columns=['Pred 0', 'Pred 1']))

improvement_final_df = pd.DataFrame(improvement_final_rows).sort_values('Macro F1', ascending=False)
print('\n=== Perbandingan improvement pada final validation ===')
display(improvement_final_df)

# Simpan terpisah dari baseline. Jangan pakai final validation untuk mencoba grid berikutnya.
# improvement_models['KNN'], improvement_models['CART'], dst. tersedia untuk saving.
```

```python
expected_models = {
    "KNN",
    "CART",
    "Logistic Regression",
    "SVM",
    "Naive Bayes"
}

assert expected_models == set(improvement_models)
assert expected_models == set(improvement_best_params)

print("Seluruh model improvement tersedia.")
display(improvement_final_df)
```

```python
y_tr = np.asarray(y_train).ravel()
y_va = np.asarray(y_val).ravel()

benchmark_rows = []

def add_comparison(name, sklearn_model, X_tr, X_va):
    sklearn_model.fit(X_tr, y_tr)
    sk_pred = sklearn_model.predict(X_va)

    scratch_pred = np.asarray(
        baseline_predictions[name]
    ).ravel()

    benchmark_rows.append({
        "Algorithm": name,
        "Scratch Accuracy": accuracy_score(y_va, scratch_pred),
        "Sklearn Accuracy": accuracy_score(y_va, sk_pred),
        "Scratch Macro F1": f1_score(
            y_va, scratch_pred, average="macro"
        ),
        "Sklearn Macro F1": f1_score(
            y_va, sk_pred, average="macro"
        )
    })


# 1. KNN
add_comparison(
    "KNN",
    KNeighborsClassifier(
        n_neighbors=5,
        metric="euclidean",
        weights="uniform",
        n_jobs=-1
    ),
    np.asarray(X_train_scaled, dtype=float),
    np.asarray(X_val_scaled, dtype=float)
)


# 2. CART
add_comparison(
    "CART",
    DecisionTreeClassifier(
        criterion="gini",
        max_depth=6,
        min_samples_split=10,
        min_samples_leaf=4,
        random_state=CFG.SEED
    ),
    np.asarray(X_train_unscaled, dtype=float),
    np.asarray(X_val_unscaled, dtype=float)
)


# 3. Logistic Regression
add_comparison(
    "Logistic Regression",
    LogisticRegression(
        penalty=None,
        solver="lbfgs",
        max_iter=2000,
        random_state=CFG.SEED
    ),
    np.asarray(X_train_scaled, dtype=float),
    np.asarray(X_val_scaled, dtype=float)
)


# 4. SVM Linear
add_comparison(
    "SVM",
    LinearSVC(
        C=1.0,
        max_iter=5000,
        random_state=CFG.SEED
    ),
    np.asarray(X_train_scaled, dtype=float),
    np.asarray(X_val_scaled, dtype=float)
)


# 5. Categorical Naive Bayes
# Bangun mapping kategori berdasarkan TRAINING saja.
nb_maps = {}
nb_train_encoded = pd.DataFrame(index=X_train_nb.index)
nb_val_encoded = pd.DataFrame(index=X_val_nb.index)

min_categories = []

for col in X_train_nb.columns:
    categories = sorted(
        X_train_nb[col].astype(str).unique()
    )

    mapping = {
        value: i for i, value in enumerate(categories)
    }

    nb_maps[col] = mapping
    unknown_code = len(categories)

    nb_train_encoded[col] = (
        X_train_nb[col].astype(str)
        .map(mapping).astype(int)
    )

    nb_val_encoded[col] = (
        X_val_nb[col].astype(str)
        .map(mapping)
        .fillna(unknown_code)
        .astype(int)
    )

    # Sediakan satu kategori tambahan untuk unknown.
    min_categories.append(len(categories) + 1)

add_comparison(
    "Naive Bayes",
    CategoricalNB(
        alpha=1.0,
        min_categories=min_categories
    ),
    nb_train_encoded,
    nb_val_encoded
)

benchmark_df = pd.DataFrame(benchmark_rows)

print("=== Baseline Scratch vs scikit-learn ===")
display(benchmark_df.round(4))
```

# Saving and Loading Models <a name="9"></a>

---

<!-- artifacts: <model>_v3.pkl -->

This section stores the trained models and restores them for prediction.

```python
# ------------------------------------------------------------
# 1. Setup direktori
# ------------------------------------------------------------
MODEL_DIR = Path(CFG.OUTPUT_DIR) / "models"
MODEL_DIR.mkdir(parents=True, exist_ok=True)

EXPECTED_MODELS = {
    "KNN",
    "CART",
    "Logistic Regression",
    "SVM",
    "Naive Bayes"
}

# Pastikan improvement sudah selesai dijalankan
assert "improvement_models" in globals(), (
    "Jalankan Model Improvement terlebih dahulu."
)

assert set(improvement_models.keys()) == EXPECTED_MODELS, (
    "Model improvement belum lengkap. "
    f"Model tersedia: {list(improvement_models.keys())}"
)

assert set(improvement_best_params.keys()) == EXPECTED_MODELS

print("Seluruh model improvement tersedia.")


# ------------------------------------------------------------
# 2. Simpan kelima model hasil improvement
# ------------------------------------------------------------
saved_model_paths = {}

for name, model in improvement_models.items():

    filename = (
        name.lower().replace(" ", "_") + ".pkl"
    )

    file_path = MODEL_DIR / filename

    with file_path.open("wb") as f:
        pickle.dump(
            model,
            f,
            protocol=pickle.HIGHEST_PROTOCOL
        )

    saved_model_paths[name] = file_path

    print(f"Model {name} disimpan: {file_path.name}")


# ------------------------------------------------------------
# 3. Simpan preprocessing bundle
# ------------------------------------------------------------
# Bundle berisi objek preprocessing yang sudah fitted
# pada training development, bukan seluruh data berlabel.

preprocessing_bundle = {

    # Identitas eksperimen
    "seed": CFG.SEED,
    "target_col": CFG.TARGET,
    "id_col": "id",

    # Cleaning
    "all_missing_cols": list(all_missing_cols),
    "drop_cols": list(drop_cols),

    # Imputasi
    "numeric_imputer": numeric_imputer,
    "categorical_imputer": categorical_imputer,
    "num_cols_to_impute": list(num_cols_to_impute),
    "cat_cols_to_impute": list(cat_cols_to_impute),

    # Feature engineering
    "mode_dow": mode_dow,
    "log_cols": list(LOG_COLS),

    # Kelompok fitur
    "numeric_features": list(NUM_CONTINUOUS),
    "binary_features": list(BINARY_FEATURES),
    "categorical_features": list(CATEGORICAL_FEATURES),

    # Urutan fitur
    "train_feature_columns": list(X_train.columns),
    "nb_feature_columns": list(X_train_nb.columns),

    # Preprocessing scaled dan unscaled
    "preprocessor_scaled": preprocessor_scaled,
    "preprocessor_unscaled": preprocessor_unscaled,

    # Preprocessing Naive Bayes
    "nb_bin_edges": {
        col: np.asarray(edges).copy()
        for col, edges in nb_bin_edges.items()
    },
    "nb_numeric_features": list(NB_NUMERIC_FEATURES),
    "nb_count_features": list(NB_COUNT_FEATURES),
    "nb_categorical_features": list(NB_CATEGORICAL_FEATURES),

    # Parameter hasil tuning
    "best_params": {
        name: params.copy()
        for name, params in improvement_best_params.items()
    }
}

bundle_path = MODEL_DIR / "preprocessing_bundle.pkl"

with bundle_path.open("wb") as f:
    pickle.dump(
        preprocessing_bundle,
        f,
        protocol=pickle.HIGHEST_PROTOCOL
    )

print(
    f"\nPreprocessing bundle disimpan: {bundle_path.name}"
)


# ------------------------------------------------------------
# 4. Load ulang kelima model
# ------------------------------------------------------------
print("\n=== Verifikasi Save/Load Model ===")

reloaded_models = {}

for name, file_path in saved_model_paths.items():

    with file_path.open("rb") as f:
        reloaded_models[name] = pickle.load(f)

    print(f"Model {name} berhasil di-load.")


# ------------------------------------------------------------
# 5. Load ulang preprocessing bundle
# ------------------------------------------------------------
with bundle_path.open("rb") as f:
    reloaded_bundle = pickle.load(f)

# Validasi metadata
assert reloaded_bundle["seed"] == CFG.SEED

assert (
    reloaded_bundle["train_feature_columns"]
    == list(X_train.columns)
)

assert (
    reloaded_bundle["nb_feature_columns"]
    == list(X_train_nb.columns)
)

assert (
    reloaded_bundle["best_params"]
    == improvement_best_params
)

print("\nMetadata preprocessing bundle valid.")


# ------------------------------------------------------------
# 6. Verifikasi hasil transformasi setelah reload
# ------------------------------------------------------------
# X_val sudah melalui cleaning, imputasi,
# feature engineering, dan log transformation.

X_val_scaled_reload = np.asarray(
    reloaded_bundle["preprocessor_scaled"].transform(X_val),
    dtype=np.float64
)

X_val_unscaled_reload = np.asarray(
    reloaded_bundle["preprocessor_unscaled"].transform(X_val),
    dtype=np.float64
)

assert np.allclose(
    X_val_scaled_reload,
    np.asarray(X_val_scaled, dtype=np.float64),
    rtol=1e-10,
    atol=1e-10
), "Scaled preprocessing berubah setelah reload!"

assert np.allclose(
    X_val_unscaled_reload,
    np.asarray(X_val_unscaled, dtype=np.float64),
    rtol=1e-10,
    atol=1e-10
), "Unscaled preprocessing berubah setelah reload!"

print("Scaled dan unscaled preprocessing identik.")


# ------------------------------------------------------------
# 7. Verifikasi diskretisasi Naive Bayes
# ------------------------------------------------------------
nb_numeric_reload = apply_nb_discretization(
    X_val_nb_raw,
    reloaded_bundle["nb_numeric_features"],
    reloaded_bundle["nb_bin_edges"]
)

nb_count_reload = apply_count_discretization(
    X_val_nb_raw,
    reloaded_bundle["nb_count_features"]
)

nb_cat_reload = X_val_nb_raw[
    reloaded_bundle["nb_categorical_features"]
].copy()

X_val_nb_reload = combine_nb_features(
    nb_numeric_reload,
    nb_count_reload,
    nb_cat_reload
)

X_val_nb_reload = X_val_nb_reload.reindex(
    columns=reloaded_bundle["nb_feature_columns"]
)

assert X_val_nb_reload.equals(X_val_nb), (
    "Preprocessing Naive Bayes berubah setelah reload!"
)

print("Naive Bayes preprocessing identik.")


# ------------------------------------------------------------
# 8. Verifikasi prediksi kelima model
# ------------------------------------------------------------
validation_inputs = {
    "KNN": X_val_scaled_reload,
    "CART": X_val_unscaled_reload,
    "Logistic Regression": X_val_scaled_reload,
    "SVM": X_val_scaled_reload,
    "Naive Bayes": X_val_nb_reload
}

verification_results = []

for name, original_model in improvement_models.items():

    loaded_model = reloaded_models[name]
    X_check = validation_inputs[name]

    pred_before = np.asarray(
        original_model.predict(X_check)
    ).ravel()

    pred_after = np.asarray(
        loaded_model.predict(X_check)
    ).ravel()

    identical = np.array_equal(
        pred_before,
        pred_after
    )

    verification_results.append({
        "Model": name,
        "Identical Predictions": identical,
        "Validation Samples": len(pred_after)
    })

    if not identical:
        raise AssertionError(
            f"Prediksi model {name} berubah setelah reload!"
        )

    print(f"{name}: Prediksi identik.")


# ------------------------------------------------------------
# 9. Ringkasan
# ------------------------------------------------------------
verification_df = pd.DataFrame(
    verification_results
)

display(verification_df)

assert verification_df["Identical Predictions"].all()

print("\n=== SAVE/LOAD SELESAI ===")
print(f"Jumlah model tersimpan: {len(saved_model_paths)}")
print(f"Direktori: {MODEL_DIR}")
print("Seluruh verifikasi prediksi berhasil.")
```

# Kaggle Submission <a name="10"></a>

---

The final pipeline and model are fitted on the original training set, and the test set is only transformed. The submission file contains the `id` and `is_canceled` columns.

```python
sample_path = CFG.DATA_DIR / "sample_submission.csv"

sample = pd.read_csv(
    sample_path,
    dtype={"id": str}
)

assert sample.columns.tolist() == ["id", CFG.TARGET]
assert len(sample) == len(test_df)
assert sample["id"].equals(test_df["id"].astype(str))

test_inputs = {
    "KNN": np.asarray(X_test_scaled, dtype=float),
    "CART": np.asarray(X_test_unscaled, dtype=float),
    "Logistic Regression": np.asarray(X_test_scaled, dtype=float),
    "SVM": np.asarray(X_test_scaled, dtype=float),
    "Naive Bayes": X_test_nb
}

assert set(improvement_models) == set(test_inputs)

CFG.OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

candidate_rows = []

for name, model in improvement_models.items():

    X_input = test_inputs[name]

    prediction = np.asarray(
        model.predict(X_input)
    ).ravel()

    assert len(prediction) == len(sample)
    assert np.isin(prediction, [0, 1]).all()

    submission = sample[["id"]].copy()
    submission[CFG.TARGET] = prediction.astype(int)

    filename = (
        "submission_"
        + name.lower().replace(" ", "_")
        + "_candidate.csv"
    )

    output_path = CFG.OUTPUT_DIR / filename
    submission.to_csv(output_path, index=False)

    candidate_rows.append({
        "Model": name,
        "File": filename,
        "Rows": len(submission)
    })

display(pd.DataFrame(candidate_rows))

print("\nSeluruh kandidat submission berhasil dibuat.")
print("Catatan: model belum di-refit pada seluruh data.")
```
